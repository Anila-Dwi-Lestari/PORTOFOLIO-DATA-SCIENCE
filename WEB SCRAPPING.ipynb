# **2️⃣ Langkah 2**

Langkah berikutnya, pilih satu halaman web yang akan di-scrap. Di sini saya menggunakan contoh sebuah halaman artikel berita dari detik.com. Periksa terlebih dahulu halaman web tersebut, contoh di sini mengambil elemen judul berita atau headline. Periksa terlebih dahulu dengan membuka halaman web yang akan dituju, lalu klik kanan pada judul berita/headline, dan pilih inspect element. Hasil yang diperoleh pada contoh adalah judul berupa elemen h1, dan class nya adalah detail__title.

````python
# Mengimpor library yang dibutuhkan
import requests                    # Digunakan untuk mengirim HTTP request ke website
from bs4 import BeautifulSoup     # Digunakan untuk mem-parsing dan membaca struktur HTML
import pandas as pd               # Digunakan untuk mengolah data dalam bentuk tabel (DataFrame)

# URL dari halaman berita yang akan di-scraping
url = 'https://news.detik.com/internasional/d-8382337/iran-klaim-hancurkan-radar-militer-terbesar-as-di-qatar'

# Mengirim permintaan ke website dan mengambil isi halaman web
response = requests.get(url)

# Mengubah HTML dari halaman web menjadi objek BeautifulSoup
# sehingga struktur HTML dapat dibaca dan dicari elemennya
soup = BeautifulSoup(response.text, 'html.parser')

# Mencari elemen judul berita pada halaman web
# Berdasarkan hasil Inspect Element pada browser Opera,
# judul berita berada pada tag <h1> dengan class 'detail__title'
headlines = soup.find_all('h1', class_='detail__title')

# Membuat list kosong untuk menyimpan hasil judul berita yang ditemukan
news_list = []

# Melakukan perulangan untuk mengambil teks dari setiap elemen headline
for headline in headlines:
    # .text digunakan untuk mengambil isi teks dari elemen HTML
    # .strip() digunakan untuk menghapus spasi kosong di awal dan akhir teks
    news_list.append(headline.text.strip())

# Menampilkan hasil scraping berupa list judul berita
print(news_list)

# Mengubah hasil list menjadi tabel menggunakan pandas DataFrame
df = pd.DataFrame(news_list, columns=['Headline'])

# Menampilkan tabel headline berita
print(df)
````

Kode di atas bertujuan untuk melakukan **web scraping** pada halaman berita di detik.com. Pertama-tama, library `requests`, `BeautifulSoup`, dan `pandas` diimpor karena masing-masing memiliki peran penting:  

- **requests**: mengirim permintaan HTTP dan mengambil isi halaman web  
- **BeautifulSoup**: mem-parsing dan membaca struktur HTML sehingga elemen tertentu bisa diekstrak  
- **pandas**: mengolah data menjadi bentuk tabel (**DataFrame**) agar lebih mudah dianalisis  

Selanjutnya, program mengirim permintaan ke URL berita yang dituju dan menyimpan respons HTML-nya. HTML tersebut kemudian diproses menggunakan **BeautifulSoup** sehingga struktur elemen halaman dapat dibaca.  

Setelah itu, program mencari elemen **judul berita**, yang berdasarkan pemeriksaan *Inspect Element* pada browser Opera berada dalam tag `<h1>` dengan class `detail__title`. Judul berita yang ditemukan disimpan dalam list `news_list` menggunakan `.text` untuk mengambil teks dari elemen HTML dan `.strip()` untuk menghapus spasi kosong di awal dan akhir.  

List ini kemudian ditampilkan menggunakan `print(news_list)`. Terakhir, list judul berita diubah menjadi **DataFrame** menggunakan pandas, sehingga data bisa dilihat dalam bentuk tabel dengan kolom bernama **"Headline"**, dan hasil tabel tersebut ditampilkan menggunakan `print(df)`.  

Output yang dihasilkan menunjukkan judul berita yang berhasil diambil, yaitu:  
**"Iran Klaim Hancurkan Radar Militer Terbesar AS di Qatar"**

# **📖 Mengambil Keseluruhan Teks Artikel Berita**

Apabila kita ingin mengambil seluruh isi teks dari artikel berita, maka kita **tidak hanya mengambil elemen judul (`<h1>`) saja**, tetapi juga harus mengambil elemen HTML yang berisi paragraf artikel.

Pada halaman berita **detik.com**, isi artikel biasanya berada pada tag `<div>` dengan **class `detail__body-text`**, yang di dalamnya terdapat beberapa tag `<p>` sebagai paragraf artikel.

# **Langkah-langkah:**

1. **Mengambil elemen container artikel**  
   Cari elemen `<div>` dengan class `detail__body-text`.

2. **Mengambil semua tag paragraf `<p>`**  
   Dari container tersebut, ambil semua tag `<p>` yang berisi teks artikel.

3. **Menggabungkan seluruh paragraf menjadi satu teks artikel**  
   Gunakan loop atau join agar semua paragraf menjadi satu string panjang.

````python
# Mengimpor library
import requests
from bs4 import BeautifulSoup

# URL artikel berita
url = "https://news.detik.com/internasional/d-8382337/iran-klaim-hancurkan-radar-militer-terbesar-as-di-qatar"

# Mengambil halaman web
response = requests.get(url)

# Parsing HTML menggunakan BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Mengambil bagian isi artikel
article = soup.find("div", class_="detail__body-text")

# Mengambil semua paragraf dalam artikel
paragraphs = article.find_all("p")

# Menyimpan teks paragraf ke dalam list
article_text = []

for p in paragraphs:
    article_text.append(p.text.strip())

# Menggabungkan semua paragraf menjadi satu teks
full_article = "\n".join(article_text)

# Menampilkan hasil teks artikel
print(full_article)
````

# **📄 Penjelasan Mengambil Keseluruhan Teks Artikel**

Untuk mengambil seluruh teks dari sebuah artikel berita, kita perlu memproses **seluruh isi konten artikel**, bukan hanya judulnya. Pada halaman berita **detik.com**, teks artikel biasanya berada dalam elemen `<div>` dengan class **`detail__body-text`**, yang di dalamnya terdiri dari beberapa paragraf `<p>`.

Pertama, program mengirim **permintaan HTTP** ke URL artikel menggunakan library **`requests`**, sehingga Python dapat mengambil seluruh **HTML halaman web** tersebut. Setelah itu, HTML yang diperoleh diproses menggunakan **BeautifulSoup** agar struktur halaman dapat dibaca dan elemen tertentu lebih mudah ditemukan.

Dengan menggunakan kode:

` soup.find("div", class_="detail__body-text") `

program akan menargetkan **container utama artikel** yang berisi seluruh isi berita. Selanjutnya, semua tag paragraf `<p>` di dalam container tersebut diambil menggunakan fungsi:

` find_all("p") `

Langkah ini dilakukan untuk mendapatkan **setiap paragraf teks dari artikel**. Teks dari masing-masing paragraf kemudian disimpan dalam sebuah **list**, lalu digabungkan menjadi satu teks lengkap menggunakan fungsi:

` "\n".join(list) `

Hasil akhirnya adalah **seluruh isi artikel berita dalam bentuk teks lengkap**, yang dapat ditampilkan pada output program atau disimpan untuk proses analisis lebih lanjut.

Dengan cara ini, proses **web scraping menjadi lebih sistematis**, sehingga kita dapat mengekstrak informasi dari halaman web secara otomatis dan memperoleh isi artikel secara lengkap tanpa kehilangan struktur paragrafnya.

# **🙋🏻‍♀️ Jawaban Soal No 3**

3. Carilah satu halaman web lain dan lakukan scrapping, sebutkan elemen mana yang harus dituliskan untuk di-scrap pada code Anda! Beri penjelasan dan ss code serta hasil yang diperoleh.

Catatan: Pada proses ini saya menggunakan browser **Google Chrome** untuk membuka halaman website dan melihat struktur HTML menggunakan fitur **Inspect Element**.

````python
# =========================================
# WEB SCRAPING DATA BUKU DARI WEBSITE
# https://books.toscrape.com
# =========================================

# Mengimpor library yang dibutuhkan
import requests
# requests digunakan untuk mengambil halaman web dari internet

from bs4 import BeautifulSoup
# BeautifulSoup digunakan untuk membaca dan mem-parsing struktur HTML
# sehingga kita bisa mencari elemen tertentu dari halaman web

import pandas as pd
# pandas digunakan untuk mengolah data hasil scraping menjadi tabel (DataFrame)


# -----------------------------------------
# Mengambil halaman website
# -----------------------------------------

# URL website yang akan di-scrape
url = "https://books.toscrape.com/"

# Mengirim HTTP request ke website
response = requests.get(url)

# Mengubah isi halaman web (HTML) menjadi objek BeautifulSoup
# agar struktur HTML dapat dibaca oleh Python
soup = BeautifulSoup(response.text, "html.parser")


# -----------------------------------------
# Mengambil data buku dari halaman web
# -----------------------------------------

# Pada website tersebut setiap buku berada pada tag:
# <article class="product_pod">
books = soup.find_all("article", class_="product_pod")

# Membuat list kosong untuk menyimpan hasil scraping
titles = []   # untuk menyimpan judul buku
prices = []   # untuk menyimpan harga buku

# Melakukan perulangan untuk mengambil data dari 10 buku pertama
for book in books[:10]:

    # Mengambil judul buku
    # Judul berada pada tag <h3> lalu di dalamnya ada tag <a>
    title = book.find("h3").find("a")["title"]

    # Mengambil harga buku
    # Harga berada pada tag <p class="price_color">
    price = book.find("p", class_="price_color").text

    # Menyimpan hasil ke dalam list
    titles.append(title)
    prices.append(price)


# -----------------------------------------
# Mengubah hasil scraping menjadi tabel
# -----------------------------------------

# Membuat DataFrame agar data lebih rapi
df = pd.DataFrame({
    "Judul Buku": titles,
    "Harga Buku": prices
})

# Menampilkan hasil scraping
print("Hasil Web Scraping Buku dari Website Books to Scrape:\n")
print(df)
````

# **Web Scraping Data Buku dari Website Books to Scrape**

Pada tugas ini dilakukan proses **web scraping** pada website https://books.toscrape.com untuk mengambil informasi mengenai buku yang ditampilkan pada halaman website. Data yang diambil berupa **judul buku dan harga buku**. Untuk mendapatkan data tersebut, kita perlu mengetahui elemen HTML pada halaman website yang menyimpan informasi tersebut.

Proses scraping dilakukan menggunakan library **requests**, **BeautifulSoup**, dan **pandas**. Library requests digunakan untuk mengambil halaman website dari internet, BeautifulSoup digunakan untuk membaca dan menelusuri struktur HTML halaman tersebut, sedangkan pandas digunakan untuk mengolah data hasil scraping agar dapat ditampilkan dalam bentuk tabel yang lebih rapi.

---

## 1. Elemen Buku

Pada website tersebut, setiap buku berada dalam tag HTML berikut:

```
<article class="product_pod">
```

Tag `article` dengan class `product_pod` berfungsi sebagai **container atau wadah utama yang menyimpan seluruh informasi tentang satu buku**, seperti judul buku, harga, rating, dan gambar buku.

Untuk mengambil seluruh data buku yang ada pada halaman, digunakan kode Python berikut:

```
books = soup.find_all("article", class_="product_pod")
```

Perintah `find_all()` digunakan untuk mencari **semua elemen HTML yang memiliki tag `article` dengan class `product_pod`**. Hasil dari perintah ini berupa kumpulan elemen buku yang kemudian dapat diproses satu per satu menggunakan perulangan.

---

## 2. Elemen Judul Buku

Struktur HTML untuk judul buku adalah sebagai berikut:

```
<h3>
   <a title="Judul Buku">
</h3>
```

Judul buku tidak berada langsung di dalam teks tag `<h3>`, tetapi tersimpan pada **atribut `title` di dalam tag `<a>` yang berada di dalam `<h3>`**. Oleh karena itu, untuk mengambil judul buku kita perlu mengakses atribut tersebut.

Kode Python yang digunakan untuk mengambil judul buku adalah:

```
title = book.find("h3").find("a")["title"]
```

Penjelasan kode:

* `find("h3")` digunakan untuk mencari tag `<h3>` pada setiap blok buku.
* `find("a")` digunakan untuk mencari tag `<a>` yang berada di dalam `<h3>`.
* `["title"]` digunakan untuk mengambil nilai dari atribut `title`, yang berisi **nama atau judul buku**.

Dengan cara ini, program dapat mengambil judul buku dari setiap data buku yang ditemukan pada halaman website.

---

## 3. Elemen Harga Buku

Harga buku pada halaman website berada dalam tag HTML berikut:

```
<p class="price_color">
```

Tag `<p>` dengan class `price_color` digunakan untuk menampilkan **harga dari setiap buku yang tersedia pada website**.

Kode Python yang digunakan untuk mengambil harga buku adalah:

```
price = book.find("p", class_="price_color").text
```

Penjelasan kode:

* `find("p", class_="price_color")` digunakan untuk mencari tag `<p>` dengan class `price_color`.
* `.text` digunakan untuk mengambil **isi teks dari elemen tersebut**, yaitu harga buku.

Dengan menggunakan kode ini, program dapat mengambil harga dari setiap buku yang terdapat pada halaman website.

Data hasil scraping kemudian disusun ke dalam **DataFrame menggunakan pandas**, sehingga tampil dalam bentuk tabel yang rapi dan mudah dibaca. Hal ini menunjukkan bahwa proses web scraping berjalan dengan baik dan informasi yang dibutuhkan dapat diperoleh secara otomatis dan terstruktur.
